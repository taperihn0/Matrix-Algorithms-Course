```python
import pandas as pd
import matplotlib
#matplotlib.use("Agg")
%matplotlib inline
import matplotlib.pyplot as plt
from pathlib import Path

base_path = Path("/Users/kajetanserwecinski/Documents/Programming/Matrix-Algorithms-Course/lab1")

def load_time_series(name: str) -> pd.DataFrame:
    return pd.read_csv(
        base_path / "stats" / "time" / f"{name}.out",
        header=None,
        names=["size", "time_ms"],
    )


def load_op_series(name: str) -> pd.DataFrame:
    op_df = pd.read_csv(
        base_path / "stats" / "op" / f"{name}.out",
        header=None,
        names=["add", "sub", "mul", "div"],
    )
    sizes = load_time_series("standart")["size"].iloc[: len(op_df)].to_numpy()
    op_df.insert(0, "size", sizes)
    return op_df


time_data = {
    name: load_time_series(name)
    for name in ["standart", "binet", "strassen", "laderman"]
}

op_data = {
    name: load_op_series(name)
    for name in ["standart", "binet", "strassen", "laderman"]
}
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

for name, df in time_data.items():
    label = {
        "standart": "Standard",
        "binet": "Binet",
        "strassen": "Strassen",
        "laderman": "Laderman",
    }[name]
    ax.plot(
        df["size"],
        df["time_ms"],
        marker="o",
        linewidth=2,
        markersize=4,
        label=label,
    )

ax.set_title("Porównanie czasu wykonania algorytmów mnożenia macierzy")
ax.set_xlabel("Rozmiar macierzy n (n x n)")
ax.set_ylabel("Czas wykonania [ms]")
ax.grid(True, linestyle="--", alpha=0.6)
ax.legend()
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10), sharex=True)
axes = axes.flatten()

for ax, name in zip(axes, ["standart", "binet", "strassen", "laderman"]):
    df = op_data[name]
    label_name = {
        "standart": "Standard",
        "binet": "Binet",
        "strassen": "Strassen",
        "laderman": "Laderman",
    }[name]

    ax.plot(df["size"], df["add"], marker="o", label="dodawanie")
    ax.plot(df["size"], df["sub"], marker="s", label="odejmowanie")
    ax.plot(df["size"], df["mul"], marker="^", label="mnożenie")
    ax.plot(df["size"], df["div"], marker="D", label="dzielenie")

    ax.set_title(f"{label_name} — liczba operacji względem rozmiaru macierzy")
    ax.set_xlabel("Rozmiar macierzy n (n x n)")
    ax.set_ylabel("Liczba operacji")
    ax.grid(True, linestyle="--", alpha=0.5)
    ax.legend(loc="best")

plt.tight_layout()
plt.show()
```
